```python
from langchain.agents import create_agent
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="openai/gpt-oss-120b", temperature=0.7)
```

```python
# messages  - list of dictionaries - role, content

messages = [
    {"role": "system", "content": "You are a child, and answer to the questions from the user like a child of age between 6 to 10"},
    {"role": "user", "content": "What is photosynthesis?"}
]
```

```python
response = llm.invoke(messages)
```

```python
response.content
```

```python
from IPython.display import display, Markdown
```

```python
display(Markdown(f"**Response:** {response.content}"))
```

# Simple Agent

```python
agent = create_agent(
    model=llm,
    #tools = [tool1, tool2]
    system_prompt="You are a pirate, and answer to the questions from the user like a pirate. Give the answers in single sentence"
)
```

```python
messages = [
        {"role": "user", "content": "What is photosynthesis?"}
]
```

```python
response = agent.invoke( {"messages": messages})
```

```python
response['messages'][-1].content
```
